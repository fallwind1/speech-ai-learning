```python
#将JUnit XML转换为CSV
import csv
import xml.etree.ElementTree as ET
from pathlib import Path

xml_path = Path(r"D:\Projects\speech-ai-learning\results\w08d7_ctc_junit.xml")
csv_path = Path(r"D:\Projects\speech-ai-learning\results\w08d7_ctc_test_table.csv")

root = ET.parse(xml_path).getroot()

rows = []

for case in root.iter("testcase"):

    failure = case.find("failure")
    error = case.find("error")
    skipped = case.find("skipped")

    if failure is not None:
        status = "FAIL"
        detail = failure.attrib.get("message", "") + "\n" + (
            failure.text or ""
        )

    elif error is not None:
        status = "ERROR"
        detail = error.attrib.get("message", "") + "\n" + (
            error.text or ""
        )

    elif skipped is not None:
        status = "SKIP"
        detail = skipped.attrib.get("message", "")

    else:
        status = "PASS"
        detail = ""

    rows.append({
        "test_name": case.attrib.get("name", ""),
        "module": case.attrib.get("classname", ""),
        "duration_sec": case.attrib.get("time", ""),
        "status": status,
        "failure_detail": detail,
    })

with open(csv_path, "w", newline="", encoding="utf-8-sig") as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "test_name",
            "module",
            "duration_sec",
            "status",
            "failure_detail",
        ],
    )

    writer.writeheader()
    writer.writerows(rows)

print("测试数量:", len(rows))
print("PASS:", sum(r["status"] == "PASS" for r in rows))
print("FAIL:", sum(r["status"] == "FAIL" for r in rows))
print("ERROR:", sum(r["status"] == "ERROR" for r in rows))
print("CSV:", csv_path)
```
